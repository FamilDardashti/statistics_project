```python
import numpy as np

mri_regions = pd.DataFrame()
mri_regions["NACCID"] = df["NACCID"]

regions = [
    "ENTORHINAL",
    "PARAHIPPOCAMPAL",
    "FUSIFORM",
    "MIDDLETEMPORAL",
    "INFERIORTEMPORAL",
    "PRECUNEUS",
    "POSTERIORCINGULATE",
    "INFERIORPARIETAL"
]

for region in regions:
    left = df[f"LH_{region}_GVOL"].replace(0, np.nan)
    right = df[f"RH_{region}_GVOL"].replace(0, np.nan)

    mri_regions[region] = (left + right) / 2
```

```python
mri_thickness = pd.DataFrame()
mri_thickness["NACCID"] = df["NACCID"]

for region in regions:
    left = df[f"LH_{region}_AVGTH"].replace(0, np.nan)
    right = df[f"RH_{region}_AVGTH"].replace(0, np.nan)

    mri_thickness[region] = (left + right) / 2
```
